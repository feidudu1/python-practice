```python
import torch
from torch.utils.data import Dataset, DataLoader
import numpy as np
import pandas as pd
import os
import cv2 as cv
import torchvision
```

# DataLoader

批量处理（batch）：如每次处理4个样本

数据打乱（shuffle）：训练时打乱数据，避免模型记住顺序；验证/测试时不需要打乱

多进程加载（num_workers）

自动转换为张量

```python
# 继承自 Dataset
class MyDataset(Dataset):
  def __init__(self, ):
    self.x = [i for i in range(10)]
    self.y = [i*2 for i in range(10)]

  def __len__(self):
    return len(self.x)
    
  def __getitem__(self, index):
    return self.x[index], self.y[index]
  
if __name__ == '__main__':
  # 创建数据集对象
  my_dataset = MyDataset()
  #  用 DataLoader 包装数据集
  # jupyter notebook执行不了多进程，只要有num_workers字段就会报错，哪怕值为1
  my_dataloader = DataLoader(my_dataset, batch_size=2)
  # my_dataloader = DataLoader(my_dataset, batch_size=2, num_workers=1)
  for x_i, y_i in my_dataloader:
    print(x_i, y_i) 

```

```python
class MyImageDataset(Dataset):
  def __init__(self):
    self.image_root = '/Users/yafei/practice/python-practice/基础学习/pytorch/tensor/animals'
    self.file_path_list = []
    self.labels = []
    dir_name = []
    for root, dirs, files in os.walk(self.image_root):
      if dirs:
        dir_name = dirs
      for file_i in files:
        file_i_full_path = os.path.join(root, file_i)
        self.file_path_list.append(file_i_full_path)
        
        label = root.split(os.sep)[-1]  # os.sep表示'/'
        label_id = dir_name.index(label)
        self.labels.append(label_id)
        
  def __len__(self):
    return len(self.file_path_list)
  
  def __getitem__(self, item):
    img = cv.imread(self.file_path_list[item])
    img = cv.resize(img, (256, 256))
    # HWC 默认是
    #tensor里的格式是 CHW
    img = np.transpose(img, (2, 0, 1))
    img_tensor = torch.from_numpy(img)
    label = self.labels[item]
    return img_tensor, label
  
if __name__ == '__main__':
  my_image_dataset = MyImageDataset()
  my_dataloader = DataLoader(my_image_dataset, shuffle=True)
  for tensor_data, label_id in my_dataloader:
    print(tensor_data.shape, label_id)
  
```

```python
import torch
import numpy as np
```

# 一、索引

```python
tensor = torch.randn(4, 4)

print(tensor)

print(f'第一行：{tensor[0]}')
print(f'第一列：{tensor[:, 0]}')
print(f'最后一列：{tensor[..., -1]}')
print(f'最后一列2：{tensor[:, -1]}')

tensor[:,1] = 0
print(tensor)
```

# 二、运算
## 1、加减乘除

```python
tensor1 = torch.tensor([
  [1, 2], [3, 4]
])
tensor2 = torch.tensor([
  [5, 6], [7, 8]
])

# 加法
print(tensor1 + tensor2)
# 讲所有元素求和
sum = tensor1.sum()
print(sum)
print(sum.item())

# 减法
print('减法')
print(tensor1 - tensor2)

# 乘法
print('乘法')
# 矩阵乘法
print(tensor1 @ tensor2)
# element-wise 乘法：类似矩阵加法，将每个对应元素相乘
print(tensor1 * tensor2)

# 除法
print('除法')
print(tensor1 / tensor2)

# 幂运算
print('幂运算')
print(tensor1 ** tensor2)


```

## 2、拼接

torch.cat(tensors, dim=0, out=None)

参数说明：

tensors (必需): 要连接的张量序列，可以是元组或列表

dim (可选): 连接的维度，默认为 0

out (可选): 输出张量

```python
tensor1 = torch.tensor([
  [1, 2], [3, 4]
])
tensor2 = torch.tensor([
  [5, 6], [7, 8]
])

# dim = 0， 沿行方向连接
print(torch.cat((tensor1, tensor2), dim=0))
# dim = 1， 沿列方向连接
print(torch.cat((tensor1, tensor2), dim=1))


```

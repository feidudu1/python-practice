## torch.randn()

torch.randn()：生成服从标准正态分布（均值 0，标准差 1）的随机数

(3, 224, 224)：张量的形状（维度）

形状含义：

3：通道数（通常表示 RGB 的 3 个通道）

224：高度（像素）

224：宽度（像素）

```python
tensor_image = torch.randn((3, 224, 224))
print(tensor_image)
transformed_image = transforms.ToPILImage()(tensor_image)
save_path = '../数学/img/transformed_image.png'
transformed_image.save(save_path)









```

# pytorch 处理图片
模拟从硬盘读取一张图片，使用pytorch在显卡上进行运算，然后把运算结果保存到硬盘

```python
image_path = '../数学/img/origin.png'
image = Image.open(image_path)

# 会自动归一化：将0-255之间的数变成0到1
transform = transforms.ToTensor()
tensor_image = transform(image)
print(tensor_image.device)

gput_tenser_image = tensor_image.to('mps')
print(gput_tenser_image.device)

gput_tenser_image += 0.2
print(gput_tenser_image)

pil_image = transforms.ToPILImage()(gput_tenser_image)
pil_image.save('../数学/img/transformed_image_add.png')





```
